```python
from Training_utils import train_loader, train
import torch_geometric as tg
import torch
from SUPG_prediction_models import *

from Training_utils import *
from torch_geometric import utils
from PyTorch_SciPy_interface import PyTorch_SciPy_interface


class mha(torch.nn.Module):
    def __init__(self):
        super().__init__()
        self.gatv2 = model = tg.nn.models.GAT(
            in_channels=7,
            hidden_channels=5,
            num_layers=3,
            out_channels=1, 
            v2 = True,
            edge_dim=2,
            add_self_loops=False
        )
    def forward(self, data) -> torch.Tensor:
        x, edge_index, edge_attr  = data.x, data.edge_index, data.edge_attr
        h = self.gatv2(
            x=x,
            edge_index=edge_index,
            edge_attr=edge_attr
        )
        return torch.abs(h)



model = mha()
optimizer = torch.optim.Adam(model.parameters(), lr=0.001)
scheduler = torch.optim.lr_scheduler.ReduceLROnPlateau(optimizer=optimizer)
    



loss_fn = torch.nn.MSELoss()


tset_dir = "training_set_v2"
tset = graph_dataset(f"data/{tset_dir}/input_values")
loader = train_loader(batch_size=15, set=tset)



curr_loss = 10
```

```python

for i in range(1000):
    loss = train(model=model, loader=loader, optimizer=optimizer, device='cpu')
    if curr_loss > loss:    
        print(f"iteration {i}: new loss: {loss}")
        curr_loss = loss
        torch.save({'model_state': model.state_dict(), 'optimizer_state': optimizer.state_dict(), 'loss': loss}, "data/models/mha_supervised_2.pth")
        #scheduler.step(loss)
    else:
        print(f"iteration {i}: {loss}")
        scheduler.step(loss)
```
