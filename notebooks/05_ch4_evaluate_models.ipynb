## 3. Read the comparison as evidence, not a leaderboard

The submitted experiments support three claims:

1. The local MLP produced the best individual results on several SPDEs, but also more unusable predictions.
2. GATv2 produced fewer unusable cases and had the best overall generalization.
3. Loss values alone did not reliably identify physical solutions, especially across different objectives and mesh configurations.

The useful conclusion is that adjacent-cell information can help on a heterogeneous dataset—not that GATv2 wins every case or solves parameter selection. The illustrated account is in `docs/results.md`; the original fields, line cuts, and exploratory comparisons remain in `archive/chapter4/Test_set_analysis.ipynb`.
